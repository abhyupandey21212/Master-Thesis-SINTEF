```python
%load_ext autoreload
import numpy as np
from utils.data_loader import *
import pandas as pd
```

```python
nodes = np.loadtxt('data/core_regions/nodes.txt')
nodes_df = pd.DataFrame(nodes, columns=['X','Y','Z'])
```

```python
c1_core_df = pd.read_csv('data/core_regions/c1_core.csv')
c1_core = c1_core_df.merge(nodes_df, on=['X', 'Y', 'Z'], how='inner').to_numpy()
c2_core_df = pd.read_csv('data/core_regions/c2_core.csv')
c2_core = c2_core_df.merge(nodes_df, on=['X', 'Y', 'Z'], how='inner').to_numpy()
sus_df = pd.read_csv('data/nodes/sus.csv')
sus = sus_df.merge(nodes_df, on=['X', 'Y', 'Z'], how='inner').to_numpy()
```

```python
c1_core.shape, c2_core.shape, sus.shape
```

```python
c1_picker = torch.zeros((c1_core.shape[0], nodes.shape[0]))
for i, row in enumerate(c1_core):
    j, x,y,z = row
    w = np.where(nodes[:,0] == x)[0]
    if len(w) > 1:
        w2 =  np.where(nodes[:,1] == y)[0]
        w = list(set(w) & set(w2))
    if len(w) > 1:
        w2 =  np.where(nodes[:,2] == z)[0]
        w = list(set(w) & set(w2))
    assert len(w) == 1
    c1_picker[i,w[0]] = 1
np.savetxt('data/core_regions/c1_picker.txt', c1_picker)
```

```python
c2_picker = torch.zeros((c2_core.shape[0], nodes.shape[0]))
for i, row in enumerate(c2_core):
    j,x,y,z = row
    w = np.where(nodes[:,0] == x)[0]
    if len(w) > 1:
        w2 =  np.where(nodes[:,1] == y)[0]
        w = list(set(w) & set(w2))
    if len(w) > 1:
        w2 =  np.where(nodes[:,2] == z)[0]
        w = list(set(w) & set(w2))
    assert len(w) == 1, (w, i)
    c2_picker[i,w[0]] = 1
np.savetxt('data/core_regions/c2_picker.txt', c2_picker)
```

```python
sus_picker = torch.zeros((sus.shape[0], nodes.shape[0]))
for i, row in enumerate(sus):
    j, x,y,z = row
    w = np.where(nodes[:,0] == x)[0]
    if len(w) > 1:
        w2 =  np.where(nodes[:,1] == y)[0]
        w = list(set(w) & set(w2))
    if len(w) > 1:
        w2 =  np.where(nodes[:,2] == z)[0]
        w = list(set(w) & set(w2))
    assert len(w) == 1
    sus_picker[i,w[0]] = 1
np.savetxt('data/core_regions/sus_picker.txt', sus_picker)
```
